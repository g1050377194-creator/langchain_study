```python
from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv
from langchain_core.prompts import message

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv('DEEPSEEK_API_KEY')
DEEPSEEK_BASE_URL = os.getenv('DEEPSEEK_BASE_URL')

model = init_chat_model(
    model="deepseek-v4-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)


print(model.invoke('你是谁').content)
```

# 字典列表

```python
message = [
    {"role": 'system', "content": '你是一个非常细心的心理医生'},
    {"role": "human", "content": "我刚刚失恋，心理着急着找下一个女性朋友是不是有问题"},
]
print(model.invoke(message).content)
```

## 02.关于多轮对话

```python
message = [
    {"role": 'system', "content": '你是一个非常细心的数学老师'},
    {"role": "human", "content": "1+2等于几"},
    {"role": "assistant", "content": "3"},
    {"role": "human", "content": "我刚刚问的什么问题"},
]
print(model.invoke(message).content)
```

## 03.ai是否拥有短期记忆

```python
message = [
    {"role":"system","content":"你是个耐心的聊天对象"},
    {"role":"human","content":"你好，我叫miki"}
]
print('对话1', model.invoke(message).content)

message1 = [{
    "role": "human","content": "我叫什么？"
}]
print('对话2', model.invoke(message1).content)
```

# 消息列表

```python
from langchain_core.messages import SystemMessage, HumanMessage

message = [
    SystemMessage("你是一个心理医生"),
    HumanMessage("我刚刚失恋，心理着急着找下一个女性朋友是不是有问题")
]

print(model.invoke(message).content)
```

## 记忆连贯性

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

message = [
    SystemMessage(content="你是一个心理医生"),
    HumanMessage(content="我刚刚失恋，心理着急着找下一个女性朋友是不是有问题"),
    AIMessage(content="你真是小丑")
]

response = model.invoke(message)
print('对话1', response.content)

message.append(AIMessage(content=response.content))
message.append(HumanMessage(content='我很犹豫要不要继续找下一段感情，还是歇歇！'))
print('对话2', model.invoke(message).content)
```
