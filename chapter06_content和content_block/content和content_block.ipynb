# 输入的区别

```python
import base64

from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv

# from chapter03_model_invoke.invoke_study_01 import response
from langchain_core.messages import HumanMessage

load_dotenv(override=True)

#1. 基本配置
NODEL_NAME = "deepseek-v4-flash"
MAX_PAIRS_HISTORY = 10
EXIT_WORD = 'quit'

print(os.getenv('DEEPSEEK_API_KEY'), os.getenv('DEEPSEEK_BASE_URL'))

#2. 初始化模型
model = init_chat_model(
    model=NODEL_NAME,
    api_key=os.getenv('DEEPSEEK_API_KEY'),
    base_url=os.getenv('DEEPSEEK_BASE_URL'),
)

# 图片转成base64
def encode_image(img_path, img_type='jpg'):
    with open(img_path, 'rb') as img_file:
        return f'data:image/{img_type};base64,{base64.b64encode(img_file.read()).decode('utf-8')}'

# 图片路径
img_path = 'clothes.PNG'

# 获取base64
image_base64 = encode_image(img_path)

# response = model.invoke([
#     HumanMessage(
#         content=[{
#             'type':'text',
#             'text':'图片的内容是什么？'
#         },{
#             'type':'image',
#             'image_url':image_base64,
#         }]
#     )
# ])

response = model.invoke([
    HumanMessage(
        content_blocks=[{
            'type':'text',
            'text':'你是什么？'
        }]
    )
])
print(response)
```
