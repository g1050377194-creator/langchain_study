```python
import os
from pathlib import Path

from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

# Notebook 在子目录里时，默认找不到项目根目录的 .env，需显式指定
ENV_PATH = Path("..") / ".env"
load_dotenv(dotenv_path=ENV_PATH, override=True)

#定义模型
API_KEY = os.getenv("DEEPSEEK_API_KEY")
BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
TAVILY_API_KEY = os.getenv("TAVILY_API_KEY")
print("TAVILY_API_KEY loaded:", bool(TAVILY_API_KEY))

model_deepseek = init_chat_model(
    model="deepseek-chat",
    model_provider="deepseek",
    api_key=API_KEY,
    base_url=BASE_URL,
)
```

## 定义工具

```python
from langchain_core.tools import tool
from langchain_tavily import TavilySearch

@tool(parse_docstring=True)
def getWeather(city: str):
    """
    获取天气

    Args:
        city: 城市名称
    """
    return f"{city}天气晴朗，气温15度左右"

print(os.getenv("TAVILY_API_KEY"))


web_search = TavilySearch(
    max_results=5,
    tavily_api_key=os.getenv("TAVILY_API_KEY"),  # 来自上一个 cell 的 load_dotenv
)

```

## 创建agent

```python
from langchain.agents import create_agent
from rich import print as rprint
from langchain.agents import create_agent

agent = create_agent(
    model=model_deepseek,
    tools=[getWeather, web_search],
)

response = agent.invoke({
    "messages": [
        {"role": "user", "content": "请问北京的天气以及北京的新闻"}
    ]
})

rprint(response)
```
